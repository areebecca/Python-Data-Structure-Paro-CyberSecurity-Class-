```python
#Alist in Python is a built-in, ordered collection of items that is mutable, meaning its elements can be changed, added, or removed after creation
x = [1,2,3,4]
x
```

```python
#append, when someone wants to add an element to a list
x = [1,2,3,4]
x.append(5)
x
```

```python
#Extending the list, we use extend if we want to add another list to an existing list
x.extend(['a','b','c','d'])
x
```

```python
#Insert, if we want to add an element and ensure it is added to the index of our choice in an existing list
x.insert(0,0)
x
```

```python
#remove ,allows  only one element at a time from an existing list to be removed
x.remove(0)
x
```

```python
#POP(), if someone wants to removes an element using the index
x.pop(-1)
x
```

```python
#index, for understanding the position of an element in a list
y=['Abonyo','Rebecca','Mirriam']
y.index('Mirriam')
```

```python
#count, if we want know the number of times a particular element reocurrs in alist
z= ['a','b','c','d','c','c','d']
z.count('c')
```

```python
#sort , it sort a list from the lowest to the highest
z.sort()
z
```

```python
#reverse , it sort our data from the highest to the lowest/ it reverses the order of the previous sorting
z.reverse()
z
```

```python
# tuple,Atuple in Python is an ordered, immutable collection of elements that can be of any data type. Tuples are defined by enclosing the comma-separated items in parentheses ()
p = ('z', 'y', 'i', 'a','z','z')
p
```

```python
#count, we are in position of counting the number of times an element reoccurs in a list
p.count('z')
```

```python
#index
p.index('y')
```

```python
#Dictionary, a built-in, mutable, and ordered collection that stores data in key-value pairs. It allows you to quickly access, insert, and delete data using a unique key as an identifier
#Dictionary, creating a dictionary using dict function
data_team =dict(
                names=['Rebecca','James','Daniel','James','Keneth'],
                Age=[24,29,23,31,30],
                Position=['Data Analyst','Data Analyst','Data Scientist','Data Supervisor','GPs Supervisor']
                )
data_team
```

```python
#Dictionary, creating a dictionary using dict function
data_team =dict([
                ('names',['Rebecca','James','Daniel','James','Keneth']),
                ('Age',[24,29,23,31,30]),
                ('Position',['Data Analyst','Data Analyst','Data Scientist','Data Supervisor','GPs Supervisor'])
                ])
data_team
```

```python
#Dictionary, creating a dictionary using {}
data_team ={
            'names':['Rebecca','James','Daniel','James','Keneth'],
            'Age':[24,29,23,31,30],
            'Position':['Data Analyst','Data Analyst','Data Scientist','Data Supervisor','GPs Supervisor']
             }
data_team
```

```python
#items , gives us the items and the values in our existing dictionary
data_team.items()
```

```python
#keys,gives the variable names in our dictionary
data_team.keys()
```

```python
#values gives the values attached to the variables
data_team.values()
```

```python
#Update, adding a new variable to the existing dictionary
data_team.update({'Gender':['Female','Male','Male','Male','Male'],'Duty Station':['Kampala','Soroti','Kampala','Kampala','Makerere']})
data_team
```

```python
#converting a dictionary in to a table
import pandas as pd
df=pd.DataFrame(data_team)
df
```
